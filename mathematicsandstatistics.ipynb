```python
#Task 1:
import numpy as np

# Vectors
vector_a = np.array([2, 4, 6])
vector_b = np.array([1, 3, 5])

# Matrices
matrix_a = np.array([
    [2, 1, 3],
    [1, 0, 2],
    [4, 1, 1]
])

matrix_b = np.array([
    [1, 2, 0],
    [3, 1, 4],
    [2, 5, 1]
])

# Vector Operations
print("Vector Addition:")
print(vector_a + vector_b)

print("\nVector Subtraction:")
print(vector_a - vector_b)

print("\nDot Product:")
print(np.dot(vector_a, vector_b))

# Matrix Operations
print("\nMatrix Addition:")
print(matrix_a + matrix_b)

print("\nMatrix Multiplication:")
print(np.dot(matrix_a, matrix_b))

# Determinant
det = np.linalg.det(matrix_a)
print("\nDeterminant of Matrix A:")
print(det)

# Inverse
if det != 0:
    print("\nInverse of Matrix A:")
    print(np.linalg.inv(matrix_a))
else:
    print("\nMatrix A is singular, inverse does not exist.")

# Eigenvalues and Eigenvectors
eigenvalues, eigenvectors = np.linalg.eig(matrix_a)

print("\nEigenvalues:")
print(eigenvalues)

print("\nEigenvectors:")
print(eigenvectors)
```

```python
#Task 2:
import numpy as np
from scipy.stats import binom
import matplotlib.pyplot as plt

# -----------------------------------
# 1. Probability of Successful Sales
# -----------------------------------

sales_success = [0.65, 0.35]

print("Probability of Successful Sales:", sales_success[0])
print("Probability of Failed Sales:", sales_success[1])

# -----------------------------------
# 2. Conditional Probability
# -----------------------------------

customer_segments = ["Premium", "Regular"]

segment_purchase_probability = {
    "Premium": 0.80,
    "Regular": 0.50
}

print("\nPremium Customer Purchase Probability:")
print(segment_purchase_probability["Premium"])

print("\nRegular Customer Purchase Probability:")
print(segment_purchase_probability["Regular"])

# -----------------------------------
# 3. Bayes Theorem
# -----------------------------------

P_A = 0.40
P_B = 0.62
P_B_given_A = 0.80

P_A_given_B = (P_B_given_A * P_A) / P_B

print("\nUpdated Probability using Bayes Theorem:")
print(P_A_given_B)

# -----------------------------------
# 4. Binomial Distribution
# -----------------------------------

n_trials = 20
success_prob = 0.65

x = np.arange(0, n_trials + 1)

binomial_probabilities = binom.pmf(x, n_trials, success_prob)

print("\nBinomial Distribution Probabilities:")
print(binomial_probabilities)

plt.plot(x, binomial_probabilities, marker='o')
plt.title("Binomial Distribution")
plt.xlabel("Successful Sales")
plt.ylabel("Probability")
plt.grid(True)
plt.show()

# -----------------------------------
# 5. Normal Distribution
# -----------------------------------

normal_mean = 50
normal_std = 10

normal_data = np.random.normal(normal_mean, normal_std, 1000)

plt.hist(normal_data, bins=30)
plt.title("Normal Distribution")
plt.xlabel("Values")
plt.ylabel("Frequency")
plt.show()

# -----------------------------------
# 6. Uniform Distribution
# -----------------------------------

uniform_low = 0
uniform_high = 1

uniform_data = np.random.uniform(uniform_low, uniform_high, 1000)

plt.hist(uniform_data, bins=30)
plt.title("Uniform Distribution")
plt.xlabel("Values")
plt.ylabel("Frequency")
plt.show()
```

```python
#Task 3:
import numpy as np
import pandas as pd
import statistics as stats
from scipy.stats import skew, kurtosis

# Dataset
sales_data = [
    120, 135, 150, 145, 160,
    155, 170, 165, 180, 175,
    190, 200, 210, 195, 185,
    175, 165, 155, 145, 135
]

# Mean
mean_value = np.mean(sales_data)
print("Mean:", mean_value)

# Median
median_value = np.median(sales_data)
print("Median:", median_value)

# Mode
mode_value = stats.mode(sales_data)
print("Mode:", mode_value)

# Range
range_value = max(sales_data) - min(sales_data)
print("Range:", range_value)

# Variance
variance_value = np.var(sales_data)
print("Variance:", variance_value)

# Standard Deviation
std_value = np.std(sales_data)
print("Standard Deviation:", std_value)

# Skewness
skewness_value = skew(sales_data)
print("Skewness:", skewness_value)

# Kurtosis
kurtosis_value = kurtosis(sales_data)
print("Kurtosis:", kurtosis_value)

# Interpretation
if skewness_value > 0:
    print("Dataset is Positively Skewed")
elif skewness_value < 0:
    print("Dataset is Negatively Skewed")
else:
    print("Dataset is Symmetric")
```
